## 1. Menginstal Alat Bantu (Library)
Python membutuhkan 'alat tambahan' untuk mengerjakan tugas spesifik.
- **pandas**: Alat untuk merapikan dan mengolah data dalam bentuk tabel (seperti Excel, tapi di Python).

## 2. Memanggil Library
Setelah diinstal, alat tersebut tidak langsung menyala. Kita harus memanggilnya terlebih dahulu ke dalam lingkungan kerja kita saat ini.

```python
import pandas as pd
import yfinance as yf

print('Library berhasil dipanggil dan siap digunakan!')
```

## 3. Mengambil Data Saham dari Internet
Kita akan mengambil data historis pergerakan saham BCA dari 1 Januari 2020 hingga hari ini.
Catatan: Di Yahoo Finance, saham perusahaan Indonesia selalu diakhiri dengan `.JK` (kode untuk Jakarta).

```python
# 1. Mendefinisikan kode saham yang ingin dicari
kode_saham = 'BBCA.JK'
saham_bca = yf.Ticker(kode_saham)

# 2. Mengunduh data historis (dari 1 Jan 2020 sampai 1 Agustus 2026)
df_bca = saham_bca.history(start='2020-01-01', end='2026-08-01')

print('Proses unduh data selesai!')
```

## 4. Mengintip Data yang Diunduh
Data yang diunduh disimpan dalam sebuah tabel yang di Python disebut **DataFrame** (disingkat `df`). 
Gunakan perintah `.head()` untuk melihat 5 baris pertama dari tabel tersebut agar kita yakin datanya benar.

```python
df_bca.head()
```

## 5. Menyimpan Data ke Komputermu
Agar besok-besok kita tidak perlu internet untuk mengunduh data yang sama, mari simpan tabel ini menjadi file `.csv` (format standar yang bisa dibuka di Excel maupun Python).

```python
df_bca.to_csv('BBCA_historical_data.csv')

print('Data berhasil disimpan menjadi file BBCA_historical_data.csv di folder yang sama dengan notebook ini.')
```
